```python
data.loc[data["Delivery_Time_Minutes"]<=0,"Delivery_Time_Minutes"]=None
data["Delivery_Time_Minutes"]=data["Delivery_Time_Minutes"].fillna(data["Delivery_Time_Minutes"].median())
```
